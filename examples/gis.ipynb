```r
library("jupytergis")
```

```r
# Create a temporary copy of the france_hiking.jGIS file
tmp_file <- "./france_hiking_tmp.jGIS"
file.copy("./france_hiking.jGIS", tmp_file, overwrite = TRUE)
stopifnot(file.exists(tmp_file))

doc <- GISDocument$new(tmp_file)
doc
```

### Raster layer

```r
google_layer <- doc$add_raster_layer(
    url = "https://mt1.google.com/vt/lyrs=y&x={x}&y={y}&z={z}",
    name = "Google Satellite",
    attribution = "Google",
    opacity = 0.6
)
```

```r
# retrieve opacity
doc$with_read(function(t) {
    doc$layers$get(t, google_layer)$parameters$opacity
})
```

```r
osm_layer <- doc$add_raster_layer(
    url = "https://tile.openstreetmap.org/{z}/{x}/{y}.png",
    name = "OSM from R",
)
```

### VectorTile layer

```r
buildings <- doc$add_vectortile_layer(
    url = "https://planetarycomputer.microsoft.com/api/data/v1/vector/collections/ms-buildings/tilesets/global-footprints/tiles/{z}/{x}/{y}",
    name = "Buildings",
    attribution = "Planetary Computer",
    opacity = 1
)
```

### Geojson layer

```r
roads <- doc$add_geojson_layer(
    path = "https://raw.githubusercontent.com/nvkelso/natural-earth-vector/master/geojson/ne_10m_roads.geojson"
)
```

```r
# retrieve name
doc$with_read(function(t) {
    doc$layers$get(t, roads)$name
})
```

### Image layer

The following cell adds an image layer to the map using `doc$add_image_layer()`.  
The image is positioned by providing the four corner coordinates as longitude/latitude pairs in order around the image.

```r
image <- doc$add_image_layer(
    url = "https://maplibre.org/maplibre-gl-js/docs/assets/radar.gif",
    coordinates = list(
    c(-80.425, 46.437),
    c(-71.516, 46.437),
    c(-71.516, 37.936),
    c(-80.425, 37.936)
  )
)
```

### GeoParquet layer

Add a vector layer from a GeoParquet file. The path can be a local file
(relative to the document) or a remote URL. The file must follow the
GeoParquet 1.x specification (WKB-encoded geometry).

```r
geoparquet_layer <- doc$add_geoparquet_layer(
    path = "https://raw.githubusercontent.com/opengeospatial/geoparquet/v1.1.0/examples/example.parquet",
    name = "Countries (GeoParquet)",
    opacity = 0.8
)
```

### GeoPackage vector layer
A GeoPackage can contain several feature tables; one layer is created per table, and a list of layer ids is returned.

```r
geopackage_vector_layers <- doc$add_geopackage_vector_layer(
    path = "https://raw.githubusercontent.com/richard-thomas/ol-load-geopackage/master/examples/dist/Natural_Earth_QGIS_layers_and_styles.gpkg",
    table_names = list("Countries", "Rivers + Lake Centrelines"),
    name = "GeoPackage example",
    type = "fill",
    opacity = 0.6
)
```

### GeoPackage raster layer

```r
geopackage_raster_layer <- doc$add_geopackage_raster_layer(
  path = "https://ngageoint.github.io/GeoPackage/examples/rivers.gpkg",
  table_names = "rivers_tiles",
  name = "rivers"
)
```
